LARGEST MOUNTAIN IN A ARRAY

```python
class Solution:
    def longestMountain(self, arr: list[int]) -> int:
        # To find peak value of the mountain first
        # THE FIRST AND LAST ELEMENTS CAN'T BE PEAK
        # SO 'i' starts in the range from second to before last element
        peak = 0
        for i in range(1, len(arr)-1): 
            if arr[i-1] < arr[i] > arr[i+1]:
                left = i - 1
                right = i + 1
                while left > 0 and arr[left] > arr[left -1] :
                    left -= 1
                while right < len(arr)-1 and arr[right] > arr[right + 1]:
                    right += 1
                peak = max(peak, right - left + 1)
        return peak
```
